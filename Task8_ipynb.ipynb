# Task 8 - Simple Data Filtering

```python
# Create a collection of student records
students = [
    {"name": "Aarav", "age": 20, "department": "IT", "marks": 85},
    {"name": "Priya", "age": 21, "department": "CS", "marks": 92},
    {"name": "Rahul", "age": 19, "department": "IT", "marks": 67},
    {"name": "Sneha", "age": 22, "department": "EXTC", "marks": 78},
    {"name": "Rohan", "age": 20, "department": "CS", "marks": 95},
    {"name": "Ananya", "age": 21, "department": "IT", "marks": 74},
    {"name": "Karan", "age": 23, "department": "EXTC", "marks": 88},
    {"name": "Meera", "age": 19, "department": "CS", "marks": 81}
]

print("Student Dataset")
for student in students:
    print(student)
```

```python
# Filter students scoring 80 or above
high_scorers = []

for student in students:
    if student["marks"] >= 80:
        high_scorers.append(student)

print("Students scoring 80 or above:")
for student in high_scorers:
    print(student)
```

```python
# Filter students younger than 21
young_students = []

for student in students:
    if student["age"] < 21:
        young_students.append(student)

print("Students younger than 21:")
for student in young_students:
    print(student)
```

```python
# Filter students from the IT department
it_students = []

for student in students:
    if student["department"] == "IT":
        it_students.append(student)

print("Students in the IT department:")
for student in it_students:
    print(student)
```

```python
# Filter using multiple conditions
filtered_students = []

for student in students:
    if student["department"] == "IT" and student["marks"] >= 80:
        filtered_students.append(student)

print("IT students scoring 80 or above:")
for student in filtered_students:
    print(student)
```

```python
print("STUDENT FILTERING RESULTS")
print("-" * 55)
print(f"{'Name':<12}{'Age':<8}{'Department':<15}{'Marks':<8}")
print("-" * 55)

for student in filtered_students:
    print(
        f"{student['name']:<12}"
        f"{student['age']:<8}"
        f"{student['department']:<15}"
        f"{student['marks']:<8}"
    )

print("-" * 55)
print("Total matching students:", len(filtered_students))
```
